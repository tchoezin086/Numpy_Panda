```python
import pandas as pd

data = {
    "Emp_ID": list(range(101, 121)),
    "Name": [
        "Amit", "Priya", "Rahul", "Neha", "Vikas",
        "Anjali", "Rohit", "Sneha", "Karan", "Pooja",
        "Arjun", "Meena", "Varun", "Kavita", "Suresh",
        "Riya", "Manish", "Simran", "Deepak", "Nisha"
    ],
    "Department": [
        "IT", "HR", "IT", "Finance", "HR",
        "IT", "Sales", "Finance", "IT", "Sales",
        "HR", "IT", "Finance", "Sales", "IT",
        "HR", "Sales", "Finance", "IT", "HR"
    ],
    "Age": [
        25, 30, 28, 35, 32,
        26, 29, 31, 27, 34,
        40, 24, 36, 29, 33,
        27, 38, 30, 26, 32
    ],
    "Salary": [
        30000, 40000, 45000, 55000, 35000,
        32000, 42000, 50000, 48000, 38000,
        60000, 28000, 52000, 43000, 47000,
        33000, 58000, 49000, 31000, 36000
    ],
    "Experience": [
        2, 5, 4, 8, 6,
        3, 5, 7, 4, 6,
        12, 1, 9, 5, 7,
        3, 11, 6, 2, 5
    ],
    "City": [
        "Delhi", "Mumbai", "Delhi", "Pune", "Delhi",
        "Noida", "Mumbai", "Pune", "Delhi", "Noida",
        "Delhi", "Noida", "Mumbai", "Delhi", "Pune",
        "Mumbai", "Delhi", "Noida", "Pune", "Delhi"
    ],
    "Performance": [
        "Good", "Excellent", "Good", "Excellent", "Average",
        "Good", "Excellent", "Good", "Excellent", "Average",
        "Excellent", "Average", "Good", "Good", "Excellent",
        "Good", "Excellent", "Average", "Good", "Excellent"
    ]
}

df = pd.DataFrame(data)

print(df)
```

```python
#Display the first 5 rows.
print(df.head(5))
```

```python
#Display only the Name, Department, and Salary columns.
print(df[["Name","Department","Salary"]])
```

```python
#Add a Bonus column containing 10% of each employee's salary.
df["Bonus"]=df["Salary"]*10/100
print(df)
```

```python
#Add an Annual_Salary column.
df["Annual_Salary"]=df["Salary"]*12
print(df)
```

```python
#Remove the Performance column.
df.drop("Performance",axis=1,inplace=True)
print(df)
```

```python
#Display employees who work in the IT department.
print(df["Name"][df["Department"]=="IT"])
```

```python
#Display employees whose salary is greater than ₹40,000.
print(df["Name"][df["Salary"]>40000])
```

```python
#Display IT employees whose salary is greater than ₹35,000.
print(df[(df["Department"] == "IT") & (df["Salary"] > 35000)])
```

```python
#Display the Name, Department, and Salary of employees whose salary is greater than ₹45,000.
print(df[["Name","Department","Salary"]][df["Salary"]>45000])
```

```python
#Display employees who live in Delhi or Mumbai.
print(df["Name"][(df["City"] == "Delhi") | (df["City"] == "Mumbai")])
```

```python
#Sort employees by salary from highest to lowest.
df.sort_values(by="Salary",ascending=False,inplace=True)
print(df)
```

```python
#Sort employees by experience from lowest to highest.
df.sort_values(by="Experience",inplace=True)
print(df)
```

```python
#Count the employees in each department.
df.groupby("Department")["Emp_ID"].count()

```

```python
#Calculate the average salary for each department.
df.groupby("Department")["Salary"].mean()
```

```python
#Find the highest salary in each department.
df.groupby("Department")["Salary"].max()
```
